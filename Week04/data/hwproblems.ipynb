```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

```python
df = pd.read_csv('metabric.csv')
df.head(10)
```

```python
x = np.sort(df["Overall Survival (Months)"])
y = np.arange(1, len(x) + 1) / len(x)

plt.plot(x, y)
plt.xlabel("Overall Survival (Months)")
plt.ylabel("ECDF")
plt.show()
```

```python
sns.ecdfplot(
    data=df,
    x="Overall Survival (Months)",
    hue="Chemotherapy"
)

plt.xlabel("Overall Survival (Months)")
plt.ylabel("ECDF")
plt.show()
```
